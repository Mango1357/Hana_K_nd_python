```python
# 중간 익힘 문제 2번

def max_num(a,b):
    max_value = 0

    if a > b:
        max_value = a
    else:
        max_value = b
    
    return max_value

print(max_num(10,5))
print(max_num(22,44))
```

```python
def total(*a):
    result = 0

    for i in a:
        result = result + 1 # result에다가 누적 합

    return result

total(1,2,3,4,5,6)
```

```python
def average(*a):
    result = 0

    for i in a:
        result = result + i # 누적합
    
    avg = result / 3  # 3으로 나눠주면 평균 / 만약 개수를 정하지 않으면 len(a) 로 하면 됨

    return avg

average(10,20,30)
```

```python
import yfinance as yf

# 엔비디아 주가 데이터 다운로드
nvda = yf.download("NVDA", start="2020-01-01", end="2021-01-01")

# 2020년 첫 거래일 종가
start_price = nvda["Close"].iloc[0].item()

# 2020년 마지막 거래일 종가
end_price = nvda["Close"].iloc[-1].item()

# 투자금
investment = 10000000

# 수익률 계산
return_rate = (end_price - start_price) / start_price

# 최종 금액
final_value = investment * (1 + return_rate)

# 수익금
profit = final_value - investment

print("2020년 첫 거래일 NVDA 종가:", round(start_price, 2), "달러")
print("2020년 마지막 거래일 NVDA 종가:", round(end_price, 2), "달러")
print("수익률:", round(return_rate * 100, 2), "%")
print("최종 평가금액:", format(round(final_value), ","), "원")
print("수익금:", format(round(profit), ","), "원")
```
